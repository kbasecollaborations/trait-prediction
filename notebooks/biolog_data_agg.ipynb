```python
import pathlib

import numpy as np
import pandas as pd
```

```python
data_dir = pathlib.Path() / "../data/raw/biolog"
```

```python
def parse_cols(cols):
    id_map = dict()
    parsed_cols = []
    for col in cols:
        if col.startswith("Unnamed"):
            parsed_col = "SSO:XXXXXXXXX"
            key = parsed_col
            value = "Unknown"
        else:
            parsed_col = col.split("__")[0]
            key = parsed_col
            value = col.split("__")[1]
        parsed_cols.append(parsed_col)
        id_map[key] = value
    return id_map, parsed_cols
```

```python
rast_id_map = dict()
```

```python
raw_ch = pd.read_csv(data_dir / "a_rast_features_CH_biolog.tsv", index_col=0, sep="\t").dropna(axis=1, how="all")
ch_id_map, ch_parsed_cols = parse_cols(raw_ch.columns)
rast_id_map.update(ch_id_map)
raw_ch.columns = ch_parsed_cols
raw_ch_source = {k: "CH" for k in raw_ch.index}
raw_ch
```

```python
raw_pmi = pd.read_csv(data_dir / "a_rast_features_PMI_biolog.tsv", index_col=0, sep="\t").dropna(axis=1, how="all")
pmi_id_map, pmi_parsed_cols = parse_cols(raw_pmi.columns)
rast_id_map.update(pmi_id_map)
raw_pmi.columns = pmi_parsed_cols
raw_pmi_source = {k: "PMI" for k in raw_pmi.index}
raw_pmi
```

```python
raw_leaf = pd.read_csv(data_dir / "a_rast_features_Leaf_microbiome_biolog.tsv", index_col=0, sep="\t").dropna(axis=1, how="all")
leaf_id_map, leaf_parsed_cols = parse_cols(raw_leaf.columns)
rast_id_map.update(leaf_id_map)
raw_leaf.columns = leaf_parsed_cols
raw_leaf_source = {k: "Leaf" for k in raw_leaf.index}
raw_leaf
```

```python
# number of na in dataframe
print(raw_ch.isna().sum().sum(), raw_pmi.isna().sum().sum(), raw_leaf.isna().sum().sum())
```

```python
# genome id (row) intersections
row_intersection = set(raw_ch.index) & set(raw_pmi.index) & set(raw_leaf.index)
row_union = set(raw_ch.index) | set(raw_pmi.index) | set(raw_leaf.index)
print(len(row_intersection), len(row_union))
```

```python
# sso id (column) intersections
col_intersection = set(raw_ch.columns) & set(raw_pmi.columns) & set(raw_leaf.columns)
col_union = set(raw_ch.columns) | set(raw_pmi.columns) | set(raw_leaf.columns)
print(len(col_intersection), len(col_union))
```

```python
# Merge dataframes
raw_df = pd.concat([raw_ch, raw_pmi, raw_leaf], axis=0, join="outer").fillna(0).astype(int)
raw_df
```

```python
id_source_map = {**raw_ch_source, **raw_pmi_source, **raw_leaf_source}
```

```python
id_annotationlist_map = raw_df.apply(lambda r: set(r.index[r.ne(0)]), axis=1).to_dict()
```

## Calculate pairwise distances and jaccard index

```python
from itertools import combinations
```

```python
# get combinations of all pairs of genome ids
genome_pairs = list(combinations(raw_df.index, 2))
len(genome_pairs)
```

```python
from multiprocessing import Pool
from tqdm import tqdm
```

```python
def calculate_scores(genome1, genome2):
    source1 = id_source_map[genome1]
    source2 = id_source_map[genome2]
    annot1 = id_annotationlist_map[genome1]
    annot2 = id_annotationlist_map[genome2]
    data1 = raw_df.loc[genome1, :]
    data2 = raw_df.loc[genome2, :]
    # calculate l1 norm
    l1_norm = np.linalg.norm(data1 - data2, ord=1) / len(data1) # type: ignore
    # calculate jaccard index
    ji = len(annot1 & annot2) / len(annot1 | annot2)
    # pearson correlation
    corr = data1.corr(data2, method="pearson") # type: ignore
    data = {
        "genome1": genome1,
        "source1": source1,
        "genome2": genome2,
        "source2": source2,
        "l1_norm": l1_norm,
        "ji": ji,
        "corr": corr,
    }
    return data
```

```python
with Pool(24) as pool:
    pairwise_scores = pool.starmap(calculate_scores, genome_pairs)
```

```python
pairwise_df = pd.DataFrame(pairwise_scores).sort_values(by="l1_norm", ascending=True)
```

```python
pairwise_df.to_csv(data_dir / "pairwise_scores_rast.csv", sep=",", index=False)
```

## Create similarity matrix and dendrogram

```python
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
alt_pairwise_df = pairwise_df.copy(deep=True)
alt_pairwise_df["genome1"] = pairwise_df["genome2"]
alt_pairwise_df["genome2"] = pairwise_df["genome1"]
dup_pairwise_df = pd.concat([pairwise_df, alt_pairwise_df], axis=0, join="outer")
```

```python
similarity_df = dup_pairwise_df.pivot(index="genome1", columns="genome2", values="l1_norm").fillna(0.0)
similarity_df
```

```python
linkage_matrix = linkage(similarity_df, method="ward", metric="euclidean")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(25, 60))
den = dendrogram(linkage_matrix, labels=similarity_df.index, leaf_rotation=0, leaf_font_size=6, orientation="right")
plt.tight_layout()
```
