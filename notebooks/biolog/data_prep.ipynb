```python
import pathlib
import pandas as pd
```

```python
biolog_folder = pathlib.Path("../../data/raw/biolog/")
features_folder = biolog_folder  / "features"
phenotypes_folder = biolog_folder  / "phenotypes"

rast_features_file = features_folder / "rast-annotations.tsv"
ch_phenotype_file = phenotypes_folder / "p_CH_phenotypes.tsv"
leaf_phenotype_file = phenotypes_folder / "p_LEAF_phenotypes.tsv"
pmi_phenotype_file = phenotypes_folder / "p_PMI_phenotypes.tsv"
```

```python
from trait_prediction.utils import read_generic_features
```

```python
rast_feature_df = read_generic_features(rast_features_file, bool_conversion=False)
rast_feature_df.index = [r.rstrip('.RAST') for r in rast_feature_df.index]
rast_feature_df
```

```python
ch_phenotype_df = pd.read_csv(ch_phenotype_file, sep="\t", index_col=0).astype('Int64')
ch_phenotype_df.columns = [f"{c}--ch_biolog" for c in ch_phenotype_df.columns]
ch_phenotype_df.index = [r.lstrip("g") for r in ch_phenotype_df.index] # type: ignore
ch_phenotype_df.index.name = "genomeID"
leaf_phenotype_df = pd.read_csv(leaf_phenotype_file, sep="\t", index_col=0).astype('Int64')
leaf_phenotype_df.columns = [f"{c}--leaf_biolog" for c in leaf_phenotype_df.columns]
pmi_phenotype_df = pd.read_csv(pmi_phenotype_file, sep="\t", index_col=0).astype('Int64')
pmi_phenotype_df.columns = [f"{c}--pmi_biolog" for c in pmi_phenotype_df.columns]
```

```python
ch_phenotype_df
```

```python
processed_folder = pathlib.Path("../../data/processed/biolog/")

processed_phenotypes_folder = processed_folder / "phenotypes"
processed_features_folder = processed_folder / "features"
```

```python
ch_phenotype_output = processed_phenotypes_folder / "ch_phenotypes.tsv"
leaf_phenotype_output = processed_phenotypes_folder / "leaf_phenotypes.tsv"
pmi_phenotype_output = processed_phenotypes_folder / "pmi_phenotypes.tsv"
ch_phenotype_df.to_csv(ch_phenotype_output, index=True, sep="\t")
leaf_phenotype_df.to_csv(leaf_phenotype_output, index=True, sep="\t")
pmi_phenotype_df.to_csv(pmi_phenotype_output, index=True, sep="\t")
```

## Split feature tables for each biolog dataset

```python
def split_data(data: pd.DataFrame, rows: pd.Index) -> pd.DataFrame:
    try:
        new_data = data.loc[rows, :]
    except KeyError:
        print("Missing keys")
        com_rows = list(set(rows).intersection(set(data.index)))
        new_data = data.loc[com_rows, :]
    new_data.index.name = "genomeID"
    return new_data.loc[new_data.any(axis=1), new_data.any(axis=0)]
```

```python
ch_rast_feature_df = split_data(rast_feature_df, ch_phenotype_df.index)
```

```python
leaf_rast_feature_df = split_data(rast_feature_df, leaf_phenotype_df.index)
```

```python
pmi_rast_feature_df = split_data(rast_feature_df, pmi_phenotype_df.index)
```

```python
print(f"CH={ch_rast_feature_df.shape}, PMI={pmi_rast_feature_df.shape}, Leaf={leaf_rast_feature_df.shape}")
```

```python
rast_ch_output = processed_features_folder / "rast_ch_features.tsv"
ch_rast_feature_df.to_csv(rast_ch_output, sep="\t", index=True)
rast_leaf_output = processed_features_folder / "rast_leaf_features.tsv"
leaf_rast_feature_df.to_csv(rast_leaf_output, sep="\t", index=True)
rast_pmi_output = processed_features_folder / "rast_pmi_features.tsv"
pmi_rast_feature_df.to_csv(rast_pmi_output, sep="\t", index=True)
```
