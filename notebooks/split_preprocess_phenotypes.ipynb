# Notebook to prepare the phenotype data for training the phenotype prediction classifiers

This notebook merges the phenotype names into a common format and splits the phenotype data into individual files

Current datasets include:
1. Literature
2. AT Leaf
3. PMI

```python
from pathlib import Path
```

```python
import pandas as pd
```

```python
# PARAMETERS
INPUT_FOLDER = Path("../data/raw/phenotypes")
OUTPUT_FOLDER = Path("../data/processed/phenotypes")
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)
```

## Read the phenotype data

```python
PHENOTYPE_FOLDER = INPUT_FOLDER
index_format_func=lambda x: x.strip().split("?")[-1].removesuffix(".RAST").removesuffix(".fna").removeprefix("g")
lit_phenotype = pd.read_csv(PHENOTYPE_FOLDER / "lit_phenotypes.tsv", sep="\t", index_col=0)
lit_index = pd.Series(lit_phenotype.index)
lit_phenotype.index = lit_index.apply(index_format_func)
atleaf_phenotype = pd.read_csv(PHENOTYPE_FOLDER / "atleaf_phenotypes.tsv", sep="\t", index_col=0)
atleaf_index = pd.Series(atleaf_phenotype.index)
atleaf_phenotype.index = atleaf_index.apply(index_format_func)
pmi_phenotype = pd.read_csv(PHENOTYPE_FOLDER / "pmi_phenotypes.tsv", sep="\t", index_col=0)
pmi_index = pd.Series(pmi_phenotype.index)
pmi_phenotype.index = pmi_index.apply(index_format_func)
```

## Find the common phenotype names

```python
def lit_func(s):
    s1 = s.removeprefix("Carbon-").removeprefix("N-Acetyl-")
    s2 = s1.removeprefix("a-").removeprefix("b-").removeprefix("g-").removeprefix("L-").removeprefix("D-")
    return s2

lit_names = pd.Index([lit_func(c) for c in lit_phenotype.columns])
lit_names
```

```python
def atleaf_func(s):
    s1 = s.removeprefix("Carbon-").removeprefix("N-Acetyl-")
    s2 = s1.removeprefix("a-").removeprefix("b-").removeprefix("g-").removeprefix("L-").removeprefix("D-")
    return s2

atleaf_names = pd.Index([atleaf_func(c) for c in atleaf_phenotype.columns])
atleaf_names
```

```python
def pmi_func(s):
    s1 = s.removeprefix("Carbon-").removeprefix("2,3-").removeprefix("D,L-").removeprefix("D.L-").removeprefix("a-").removeprefix("g-").removeprefix("L-").removeprefix("D-").removeprefix("")
    s2 = s1.replace(" Acid", "-Acid")
    return s2

pmi_names = pd.Index([pmi_func(c) for c in pmi_phenotype.columns])
list(pmi_names)
```

```python
from collections import Counter

c = Counter(pmi_names)
c.most_common(20)
```

## Manual refinement

I copied these names into Excel and manually updated the names to be more consistent across datasets

```python
common_names = pd.read_csv("../data/interim/phenotype_common_names.csv")
common_names
```

```python
lit_common_names = list(common_names.loc[:, "Lit"].dropna())
len(lit_common_names)
```

```python
atleaf_common_names = list(common_names.loc[:, "AT Leaf"].dropna())
len(atleaf_common_names)
```

```python
pmi_common_names = [n.replace(" ", "-") for n in common_names.loc[:, "PMI"].dropna()]
len(pmi_common_names)
```

## Writing the phenotype data

```python
from fuzzywuzzy import fuzz
```

```python
def write_phenotype_files(phenotype, common_names, output_folder):
    output_folder.mkdir(parents=True, exist_ok=True)
    for name, common_name in zip(phenotype.columns, common_names):
        if fuzz.WRatio(name, common_name) < 90:
            print(f"Name mismatch: {name} vs {common_name}")
        s = phenotype.loc[:, name].astype("Int64")
        s.rename(common_name, inplace=True)
        s.to_csv(output_folder / f"{common_name}.tsv", sep="\t")
```

```python
write_phenotype_files(lit_phenotype, lit_common_names, OUTPUT_FOLDER / "lit")
write_phenotype_files(atleaf_phenotype, atleaf_common_names, OUTPUT_FOLDER / "atleaf")
write_phenotype_files(pmi_phenotype, pmi_common_names, OUTPUT_FOLDER / "pmi")
```
